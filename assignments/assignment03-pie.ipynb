```python
# Importing libraries
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import Extract

# Loading data
url = "https://drive.google.com/uc?id=1AWPf-pJodJKeHsARQK_RHiNsE8fjPCVK&export=download"

# Reading the CSV file directly from the URL
df = pd.read_csv(url)

# Looking at data
df.head()
```

```python
# Extracting email domain names
df['domain'] = df['email'].apply(lambda x: x.split('@')[1])
# https://stackoverflow.com/questions/53044548/how-to-extract-domain-from-email-address-with-pandas

# Counting how many times each domain appears
domain_counts = df['domain'].value_counts()

# Displaying the top few to check
domain_counts.head()
```
