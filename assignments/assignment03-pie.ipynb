```python
# Importing libraries
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import Extract

# Loading data
url = "https://drive.google.com/uc?id=1AWPf-pJodJKeHsARQK_RHiNsE8fjPCVK&export=download"

# Reading the CSV file directly from the URL
df = pd.read_csv(url)
# https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html

# Looking at data
df.head()
# https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.head.html
```

```python
# Checking column names to confirm the correct email column
print("Columns in dataset:", df.columns)
# https://www.geeksforgeeks.org/pandas/dealing-with-rows-and-columns-in-pandas-dataframe/

# Extracting email domain names
df['domain'] = df['Email'].apply(lambda x: x.split('@')[1])
# https://stackoverflow.com/questions/53044548/how-to-extract-domain-from-email-address-with-pandas

# Counting how many times each domain appears
domain_counts = df['domain'].value_counts()
# https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.value_counts.html

# Displaying the top few to check
domain_counts.head()
```

```python
# Creating a pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    domain_counts.head(6),
    labels=domain_counts.index[:6],
    autopct='%1.1f%%',
    startangle=140,
    shadow=True
)

plt.title("Distribution of Email Domains", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.figure.html
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.pie.html
# https://stackoverflow.com/questions/6170246/how-do-i-use-matplotlib-autopct
```
